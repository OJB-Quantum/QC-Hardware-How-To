## Magnetic Field Interference in Josephson Weak Links

## Acronyms and Important Symbols

| Symbol/ Acronym | Definition | Value/ SI Unit |
| --- | --- | --- |
| $I_s$ | Supercurrent flowing through the weak link | $\mathrm{A}$ |
| $I_c$ | Critical current of the Josephson weak link | $\mathrm{A}$ |
| $I_c(0)$ | Zero-field critical current | $\mathrm{A}$ |
| $\varphi$ | Gauge invariant superconducting phase difference | $\mathrm{rad}$ |
| $B$ | Applied magnetic field magnitude | $\mathrm{T}$ |
| $\Phi$ | Total magnetic flux threading the junction area | $\mathrm{Wb}$ |
| $\Phi_0$ | Superconducting flux quantum | $\approx 2.068\times10^{-15}\ \mathrm{Wb}$ |
| $h$ | Planck constant | $\approx 6.626\times10^{-34}\ \mathrm{J}\cdot\mathrm{s}$ |
| $e$ | Elementary charge | $\approx 1.602\times10^{-19}\ \mathrm{C}$ |
| $A_{\mathrm{eff}}$ | Effective magnetic area of the junction | $\mathrm{m}^2$ |
| $B_1$ | Applied magnetic field corresponding to the first ideal Fraunhofer minimum | $\mathrm{T}$ |
| $A_{\mathrm{SIS}}$ | Effective area of the conventional planar superconductor-insulator-superconductor baseline junction | $1.0\times10^{-12}\ \mathrm{m}^2$ |
| $B_{1,\mathrm{SIS}}$ | First Fraunhofer minimum magnetic field for the planar SIS junction | $\approx 2.07\ \mathrm{mT}$ |
| $A_{\mathrm{NB}}$ | Geometric area proxy assigned to nanoscale weak link candidates | $1.60\times10^{-15}\ \mathrm{m}^2$ |
| $B_{1,\mathrm{NB}}$ | First Fraunhofer minimum magnetic field for the nanobridge geometry | $\approx 1.29\ \mathrm{T}$ |
| $x$ | Dimensionless magnetic flux argument for the Fraunhofer diffraction envelope | $\mathrm{dimensionless}$ |
| SIS | Superconductor-Insulator-Superconductor | $\mathrm{junction\ type}$ |

A Josephson weak link supports a phase dependent supercurrent whose simplest current phase relation is

$$I_s = I_c \sin(\varphi),$$

where $I_c$ is the critical current and $\varphi$ is the gauge invariant superconducting phase difference across the weak link [1,2]. An applied magnetic field introduces a spatial phase gradient across a junction with finite magnetic area. For a junction carrying approximately uniform supercurrent density, integrating these phase shifted contributions produces the familiar Fraunhofer envelope [2,3]

$$I_c(B) = I_c(0) \left\vert{} \frac{\sin\left(\pi \Phi / \Phi_0\right)}{\pi \Phi / \Phi_0} \right\vert{},$$

with magnetic flux

$$\Phi = B A_{\mathrm{eff}},$$

and superconducting flux quantum

$$\Phi_0 = \frac{h}{2e} \approx 2.068\times10^{-15}\ \mathrm{Wb}.$$

The first ideal Fraunhofer minimum occurs once one flux quantum threads the effective junction area,

$$B_1 = \frac{\Phi_0}{A_{\mathrm{eff}}}.$$

The script in this notebook uses this area scaling to contrast a conventional planar SIS junction with a collection of nanoscale weak links. For the $1\ \mu\mathrm{m}^2$ planar baseline,

$$A_{\mathrm{SIS}} = 1.0\times10^{-12}\ \mathrm{m}^2,$$

which gives

$$B_{1,\mathrm{SIS}} \approx 2.07\ \mathrm{mT}.$$

Several zeros and secondary lobes therefore appear inside the simulated magnetic field interval.

For every nanobridge candidate, the script assigns the common geometric area proxy

$$A_{\mathrm{NB}} = (80\ \mathrm{nm})(20\ \mathrm{nm}) = 1.60\times10^{-15}\ \mathrm{m}^2.$$

Its corresponding first idealized interference minimum lies at

$$B_{1,\mathrm{NB}} = \frac{\Phi_0}{A_{\mathrm{NB}}} \approx 1.29\ \mathrm{T},$$

which is more than twenty five times larger than the maximum field magnitude plotted in the calculation. At $B=50\ \mathrm{mT}$,

$$\frac{\Phi}{\Phi_0} = \frac{BA_{\mathrm{NB}}}{\Phi_0} \approx 0.0387,$$

giving

$$\frac{I_c(50\ \mathrm{mT})}{I_c(0)} \approx 0.99754.$$

The idealized orbital interference term therefore suppresses the nanobridge critical current by only about 0.25% at the edge of the simulated field interval. This follows directly from the small argument expansion

$$\frac{\sin x}{x} \approx 1-\frac{x^2}{6}, \qquad \vert{}x\vert{}\ll1,$$

where

$$x = \frac{\pi B A_{\mathrm{eff}}}{\Phi_0}.$$

### Physical Interpretation

Within this geometric model, nanoscale confinement shifts the characteristic Fraunhofer field scale from millitesla values toward the tesla regime. The planar SIS baseline consequently develops pronounced diffraction lobes across the selected field range, whereas the nanoscale traces remain nearly flat apart from their different zero field critical current amplitudes.

The material labels in the simulation change $I_c(0)$ alone. Every non SIS candidate receives the same $A_{\mathrm{eff}}$, so its normalized magnetic response

$$\frac{I_c(B)}{I_c(0)}$$

is mathematically identical. The calculation therefore isolates a geometric scaling argument and leaves material specific electrodynamics outside the present model.

This is relevant for NbN, MoGe, granular aluminium, NbTiN, topological semimetals, and related weak link materials. Real nanobridges can exhibit magnetic field responses associated with screening currents, flux focusing, vortex entry, nonuniform current density, and material dependent current phase relations [3–5]. Experiments on narrow Nb, NbN, and TaN bridges have demonstrated field dependent critical current oscillations associated with vortex configurations [4]. Dayem bridge experiments likewise show that supercurrent can persist under sizable perpendicular magnetic fields [5].

The computation should be interpreted as demonstrating geometric displacement of the conventional Fraunhofer diffraction scale beyond the selected $\pm50\ \mathrm{mT}$ window, instead of some universal elimination of magnetic field sensitivity. The large contrast between the planar SIS junction and the nanobridge candidates arises primarily from the approximately $625$ fold reduction in assigned effective area,

$$\frac{A_{\mathrm{SIS}}}{A_{\mathrm{NB}}} = \frac{1.0\times10^{-12}}{1.60\times10^{-15}} = 625.$$

This same factor increases the modeled field required to accumulate one superconducting flux quantum.

### Conceptual Map

```text
External magnetic field B
│
├── Magnetic flux
│   └── Φ = B A_eff
│
├── Gauge invariant phase variation
│   └── Spatial interference of supercurrent
│
├── Extended planar SIS junction
│   ├── Large A_eff
│   ├── Φ / Φ0 grows rapidly
│   └── Fraunhofer minima enter the mT range
│
└── Nanoscale bridge model
    ├── Small assigned A_eff
    ├── Φ / Φ0 remains small over ±50 mT
    ├── sinc envelope stays near unity
    └── First modeled minimum moves toward the tesla range
```

Here, *Fraunhofer* refers to the mathematical analogy between Josephson phase interference and single slit optical diffraction. The term originates from Joseph von Fraunhofer, whose work on far field optical diffraction supplied the historical terminology later adopted for the characteristic sinc shaped Josephson magnetic interference envelope.

````python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------
# Plotting Parameters
# ---------------------------------------------------
plt.rcParams['figure.dpi'] = 250
plt.rcParams['font.sans-serif'] = ['Tahoma', 'DejaVu Sans']
plt.rcParams['font.weight'] = 'normal'

def simulate_comprehensive_interference() -> None:
    """This script evaluates the magnetic field resilience of topological
    semimetals, kinetic inductors, and disordered superconductors against
    a conventional planar tunneling junction baseline.
    """

    # ---------------------------------------------------
    # Control Knobs
    # ---------------------------------------------------
    magnetic_field_range = np.linspace(-50.0, 50.0, 2500)  # Magnetic field B in mT
    flux_quantum = 2.067833848e-15  # Magnetic flux quantum in Weber

    # Standardized geometric area parameters
    bridge_width_nm = 80.0
    bridge_thickness_nm = 20.0
    nanobridge_area_m2 = (bridge_width_nm * 1e-9) * (bridge_thickness_nm * 1e-9)
    conventional_sis_area_m2 = 1.0e-12  # One square micrometer

    # Characteristic critical current scales (I_star) in microamperes
    material_currents = {
        'Conventional SIS Baseline': 25.0,
        'NbN (Kinetic ScS)': 29.6,
        'MoGe (Disordered ScS)': 12.5,
        'grAl (Disordered ScS)': 8.9,
        'NbTiN (Kinetic ScS)': 8.0,
        'Pt3Sn (Bulk Nodal NcN)': 4.5,
        'NiTe2 (Dirac Semimetal NcN)': 3.4,
        'Pt3SnFe (Iron Spintronic NcN)': 2.8,
        'PtSe2 (Dirac Semimetal NcN)': 2.1,
        'Cd3As2 (Topological Dirac NcN)': 1.8,
        'MoTe2 (Weyl Semimetal NcN)': 1.5,
        'WSe2 (Topological NcN)': 1.4,
        'WTe2 (Weyl Semimetal NcN)': 1.2,
        'Graphene (Monolayer NcN)': 0.3
    }

    # Effective cross-sectional areas corresponding to the geometry
    material_areas = {
        key: conventional_sis_area_m2 if 'SIS' in key else nanobridge_area_m2
        for key in material_currents.keys()
    }

    # Distinct hexadecimal color codes for visual separation
    material_colors = {
        'Conventional SIS Baseline': '#2f2f2f',
        'NbN (Kinetic ScS)': '#d62728',
        'MoGe (Disordered ScS)': '#8c564b',
        'grAl (Disordered ScS)': '#ff7f0e',
        'NbTiN (Kinetic ScS)': '#e377c2',
        'Pt3Sn (Bulk Nodal NcN)': '#1f77b4',
        'NiTe2 (Dirac Semimetal NcN)': '#17becf',
        'Pt3SnFe (Iron Spintronic NcN)': '#2ca02c',
        'PtSe2 (Dirac Semimetal NcN)': '#98df8a',
        'Cd3As2 (Topological Dirac NcN)': '#9467bd',
        'MoTe2 (Weyl Semimetal NcN)': '#c5b0d5',
        'WSe2 (Topological NcN)': '#8c6d31',
        'WTe2 (Weyl Semimetal NcN)': '#bcbd22',
        'Graphene (Monolayer NcN)': '#aec7e8'
    }

    # ---------------------------------------------------
    # Mathematical Modeling
    # ---------------------------------------------------
    def calculate_absolute_current(field_mt: np.ndarray, area: float, max_current: float) -> np.ndarray:
        """This function calculates the absolute critical current incorporating the geometric suppression envelope."""
        field_tesla = field_mt * 1e-3
        flux = field_tesla * area
        phase_argument = (np.pi * flux) / flux_quantum

        with np.errstate(divide='ignore', invalid='ignore'):
            envelope = np.abs(np.sin(phase_argument) / phase_argument)
            envelope = np.where(phase_argument == 0, 1.0, envelope)

        return max_current * envelope

    # ---------------------------------------------------
    # Visualization Rendering
    # ---------------------------------------------------
    plt.figure(figsize=(11.5, 7.5))

    for material, baseline_current in material_currents.items():
        absolute_current = calculate_absolute_current(
            magnetic_field_range,
            material_areas[material],
            baseline_current
        )

        is_baseline = 'SIS' in material
        plt.plot(
            magnetic_field_range,
            absolute_current,
            label=f'{material} ($I_c \\approx$ {baseline_current} $\\mu$A)',
            color=material_colors[material],
            linestyle='-' if is_baseline else '-',
            linewidth=2.8 if is_baseline else 1.6,
            zorder=10 if is_baseline else 2
        )

    plt.xlabel('External Magnetic Field (mT)')
    plt.ylabel('Absolute Critical Current ($\\mu$A)')
    plt.title('Magnetic Field Resilience in Superconducting Qubit Junctions')

    plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), framealpha=0.95, fontsize=8)
    plt.grid(True, linestyle=':', alpha=0.7)

    plt.tight_layout()
    plt.show()

if __name__ == '__main__':
    simulate_comprehensive_interference()
````
