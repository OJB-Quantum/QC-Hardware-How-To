```python
"""This script visualizes the gain, bandwidth, and saturation comparison between KI-TWPA, NKPA, and J-TWPA devices."""

import matplotlib.pyplot as plt
import numpy as np

# Control Knobs
FONT_SANS_SERIF = ["DejaVu Sans"]
FONT_WEIGHT = "normal"
DPI_SETTING = 250
FIGURE_SIZE = (10, 7)

plt.rcParams["font.sans-serif"] = FONT_SANS_SERIF
plt.rcParams["font.weight"] = FONT_WEIGHT
plt.rcParams["axes.labelweight"] = FONT_WEIGHT
plt.rcParams["axes.titleweight"] = FONT_WEIGHT
plt.rcParams["figure.dpi"] = DPI_SETTING

# Device Parameters
devices = ["KI-TWPA (3WM)", "NKPA (Nanobridge)", "J-TWPA (RPM N=1)", "J-TWPA (N=4)"]
gain_db = [16.5, 24.0, 20.0, 31.0]  # Peak Gain in dB
bandwidth_ghz = [2.0, 0.0031, 3.0, 3.0]  # Instantaneous Bandwidth in GHz
p1db_dbm = [-63.0, -133.0, -103.0, -84.0]  # 1 dB Compression Power in dBm
added_noise_quanta = [3.1, 0.59, 1.0, 1.2]  # Added Noise in Quanta
bar_colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]

fig, axs = plt.subplots(2, 2, figsize=FIGURE_SIZE)
axs = axs.flatten()

# Plot 1: Peak Gain Comparison
axs[0].bar(devices, gain_db, color=bar_colors)
axs[0].set_ylabel("Peak Gain (dB)")
axs[0].set_title("Peak Amplification Gain")
axs[0].tick_params(axis="x", rotation=15)

# Plot 2: Instantaneous Bandwidth (Log Scale)
axs[1].bar(devices, bandwidth_ghz, color=bar_colors)
axs[1].set_yscale("log")
axs[1].set_ylabel("Bandwidth (GHz)")
axs[1].set_title("Instantaneous Bandwidth (Log Scale)")
axs[1].tick_params(axis="x", rotation=15)

# Plot 3: 1 dB Gain Compression Power
axs[2].bar(devices, p1db_dbm, color=bar_colors)
axs[2].set_ylabel("1 dB Compression Power (dBm)")
axs[2].set_title("Dynamic Range ($P_{-1dB}$)")
axs[2].tick_params(axis="x", rotation=15)

# Plot 4: Added Noise Performance
axs[3].bar(devices, added_noise_quanta, color=bar_colors)
axs[3].axhline(y=0.5, color="black", linestyle="--", label="Quantum Limit (0.5 quanta)")
axs[3].set_ylabel("Added Noise ($N_{add}$ Quanta)")
axs[3].set_title("Noise Performance")
axs[3].legend()
axs[3].tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.show()
```
