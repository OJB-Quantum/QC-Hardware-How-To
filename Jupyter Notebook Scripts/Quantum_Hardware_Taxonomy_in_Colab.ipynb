```python
"""
This script generates a directed graph visualizing a quantum hardware taxonomy.
"""
import graphviz
import matplotlib.pyplot as plt
from IPython.display import Image, display

# Primary Control Knobs
FIGURE_DPI = 250
PRIMARY_FONT = 'Tahoma'
BACKUP_FONT = 'DejaVu Sans'
FONT_WEIGHT = 'normal'
GRAPH_LAYOUT = 'dot'
RANK_DIRECTION = 'LR'
NODE_STYLE = 'box'
OUTPUT_FORMAT = 'png'


def configure_matplotlib_parameters() -> None:
    """
    This script function configures the global mathematical plotting parameters.
    """
    plt.rcParams['figure.dpi'] = FIGURE_DPI
    plt.rcParams['font.sans-serif'] = [PRIMARY_FONT, BACKUP_FONT]
    plt.rcParams['font.weight'] = FONT_WEIGHT


def build_graph(taxonomy_data: dict | list, parent_node: str, directed_graph: graphviz.Digraph) -> None:
    """
    This script function recursively populates the directed graph with categorized nodes and directional edges.
    """
    if isinstance(taxonomy_data, dict):
        for key, value in taxonomy_data.items():
            directed_graph.node(key, key, shape=NODE_STYLE, fontname=PRIMARY_FONT)
            directed_graph.edge(parent_node, key)
            build_graph(value, key, directed_graph)
    elif isinstance(taxonomy_data, list):
        for item in taxonomy_data:
            directed_graph.node(item, item, shape=NODE_STYLE, fontname=PRIMARY_FONT)
            directed_graph.edge(parent_node, item)


def main() -> None:
    """
    This script function orchestrates the overarching graph construction and streams the output directly to the notebook display interface.
    """
    configure_matplotlib_parameters()

    hardware_taxonomy = {
        "I. Quantum Core Hardware": {
            "A. Qubit Technologies": {
                "1. Superconducting Qubits": [
                    "Transmon, Fluxonium, Flux qubit",
                    "Cavity protected (cat, binomial, GKP encoded)"
                ],
                "2. Spin Based Qubits": {
                    "a. Semiconductor Spins (Si, SiGe, GaAs, donors, NV)": [],
                    "b. Magnetic and Molecular Spins": [
                        "Magnetic clusters (Fe8, Mn12, heterometallic rings, other candidates)",
                        "Magnetic nanodiscs (meron, skyrmion qubits)"
                    ]
                },
                "3. Bosons (microwave photons, phonons, magnons)": [],
                "4. Topological or Majorana Candidates": []
            },
            "B. Quantum Interconnects (Buses)": {
                "1. Planar Resonators (CPW lambda 4, lambda 2, lumped, stripline)": [],
                "2. 3 Dimensional Superconducting Cavities": [],
                "3. Metamaterial Waveguides and Resonators": [],
                "4. Photonic Waveguides and Ring resonator PICs": [],
                "5. Hybrid Quantum Transducers (electro optic, electro acoustic, magnonic)": []
            },
            "C. Quantum Limited and Quantum Enhanced Detectors": {
                "1. SNSPD": [],
                "2. KID or MKID": [],
                "3. Rubidium Photomultipliers (JPM) and Photonics": [],
                "4. Quantum optimized Bolometers or Calorimeters": []
            },
            "D. Quantum Memories": {
                "1. Rare earth AFC crystals": [],
                "2. Magnon memories": [],
                "3. 3 Dimensional Cat code cavities": [],
                "4. Nuclear spin ensembles": []
            },
            "E. Quantum Photonic Integrated Circuits (QPICs)": {
                "1. SiN, Si, SiO2 wafer scale": [],
                "2. III V hybrids (GaAs, InP)": [],
                "3. Diamond and LiNbO3": []
            }
        },
        "II. Quantum Adjacent Hardware": {
            "A. Cryogenic Digital Control Logic": {
                "1. Single Flux Quantum families (RSFQ, RQL, AQFP, eSFQ)": [],
                "2. Deep Cryo CMOS (4 K)": [],
                "3. Milli Kelvin CMOS (under 100 mK)": []
            },
            "B. Cryogenic Mixed Signal and RF ICs": {
                "1. Time interleaved DAC or ADC": [],
                "2. RF Transceiver SoCs (2 to 18 GHz I Q)": [],
                "3. Cryo Class 0 Drivers or Plexos": []
            },
            "C. Cryogenic Amplifiers, Filters, and Passive Components": {
                "1. mk Parametric Pre-Amplifiers": [
                    "a. Flux pumped Superconducting Parametric Amplifier or Converter (JPA or JPC)",
                    "b. Josephson Traveling Wave Parametric Amplifier (JTWPA)",
                    "c. Kinetic Inductance Traveling Wave Parametric Amplifier (KI TWPA)",
                    "d. Nanobridge Kinetic Parametric Amplifier (NKPA)",
                    "e. Quantum Capacitance Parametric Amplifier (QCPA)",
                    "f. SNAIL based Parametric Amplifier (SPA or SNAIL TWPA)"
                ],
                "2. 4 K HEMT LNAs (octave wide, high dynamic range)": [],
                "3. RF Isolators or Circulators (ferrite or on chip)": [],
                "4. Superconducting and SAW Filters": []
            },
            "D. Cryogenic Packaging and Interconnects": {
                "1. Flex print and interposer tiles": [],
                "2. 3 Dimensional cavities with cryogenic bump bonds or interconnects": [],
                "3. Coax, waveguide, or stripline wiring (NbTi, Nb, CuNi)": [],
                "4. Optical fiber feedthroughs (1 to 4 K)": [],
                "5. Magnetic and vibration shielding, radiation hardeners": []
            },
            "E. Cryogenic Memory and Storage": {
                "1. SRAM (FiNET4M and sn cram cryo SRAM)": [],
                "2. Floating Body RAM (FBRAM) at 77 K": [],
                "3. Capacitor less eDRAM or DRAM benchmarks (2T0C, 4 K)": [],
                "4. Jb based RAM (JJ RAM, JHRAM)": [],
                "5. Spin orbit torque (SOT) MRAM at 4 K": []
            }
        }
    }

    dot_diagram = graphviz.Digraph(
        comment='Quantum Hardware Categories',
        engine=GRAPH_LAYOUT
    )

    dot_diagram.attr(rankdir=RANK_DIRECTION, dpi=str(FIGURE_DPI))

    root_node = "Extensive Map of Quantum Hardware Categories"
    dot_diagram.node(root_node, root_node, shape=NODE_STYLE, fontname=PRIMARY_FONT)

    build_graph(hardware_taxonomy, root_node, dot_diagram)

    png_data = dot_diagram.pipe(format=OUTPUT_FORMAT)
    display(Image(data=png_data))


if __name__ == '__main__':
    main()
```
